## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

```python
# Memeriksa rentang nilai pada kolom kategorikal
valid_values = {
    'season': [1, 2, 3, 4],
    'yr': [0, 1],
    'mnth': list(range(1, 13)),
    'holiday': [0, 1],
    'weekday': list(range(7)),
    'workingday': [0, 1],
    'weathersit': [1, 2, 3, 4]
}

# Pemeriksaan dataset harian dan per jam
for name, df in [('day_df', day_df), ('hour_df', hour_df)]:
    print(f"\nPemeriksaan {name}")

    for column, allowed in valid_values.items():
        invalid = (~df[column].isin(allowed)).sum()
        print(f"{column}: {invalid} nilai tidak valid")

    if 'hr' in df.columns:
        invalid_hr = (~df['hr'].between(0, 23)).sum()
        print(f"hr: {invalid_hr} nilai tidak valid")

    # Memeriksa rentang variabel cuaca
    for column in ['temp', 'atemp', 'hum', 'windspeed']:
        invalid = (~df[column].between(0, 1)).sum()
        print(f"{column}: {invalid} nilai di luar rentang")

    # Memeriksa konsistensi jumlah penyewaan
    mismatch = (
        df['casual'] + df['registered'] != df['cnt']
    ).sum()

    print("Ketidaksesuaian jumlah penyewaan:", mismatch)
```

```python
# Fungsi untuk menghitung outlier menggunakan IQR
def check_outliers(df, columns):
    results = []

    for column in columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = (
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        ).sum()

        results.append({
            'Kolom': column,
            'Jumlah Outlier': outliers,
            'Lower Bound': round(lower_bound, 2),
            'Upper Bound': round(upper_bound, 2)
        })

    return pd.DataFrame(results)


# Kolom numerik yang akan diperiksa
numeric_columns = [
    'temp', 'atemp', 'hum', 'windspeed',
    'casual', 'registered', 'cnt'
]

# Pemeriksaan outlier dataset per jam
outlier_hour = check_outliers(hour_df, numeric_columns)

print("Hasil Pemeriksaan Outlier Dataset Per Jam:")
display(outlier_hour)

# Pemeriksaan outlier dataset harian
outlier_day = check_outliers(day_df, numeric_columns)

print("Hasil Pemeriksaan Outlier Dataset Harian:")
display(outlier_day)
```

## Exploratory Data Analysis (EDA)

### Explore ...

```python
# Membuat salinan dataset untuk analisis
hour_eda = hour_df.copy()
day_eda = day_df.copy()

# Membuat kolom tahun dan bulan
hour_eda['year'] = hour_eda['dteday'].dt.year
hour_eda['month'] = hour_eda['dteday'].dt.month
hour_eda['date'] = hour_eda['dteday'].dt.date

day_eda['year'] = day_eda['dteday'].dt.year
day_eda['month'] = day_eda['dteday'].dt.month

# Membuat kategori waktu berdasarkan jam
def categorize_hour(hour):
    if 0 <= hour <= 4:
        return 'Dini Hari'
    elif 5 <= hour <= 10:
        return 'Pagi'
    elif 11 <= hour <= 15:
        return 'Siang'
    elif 16 <= hour <= 19:
        return 'Sore'
    else:
        return 'Malam'

hour_eda['time_category'] = hour_eda['hr'].apply(categorize_hour)

# Membuat label kondisi cuaca
weather_labels = {
    1: 'Cerah / Sedikit Berawan',
    2: 'Berawan / Berkabut',
    3: 'Hujan Ringan / Salju Ringan',
    4: 'Cuaca Buruk'
}

hour_eda['weather_label'] = hour_eda['weathersit'].map(weather_labels)
day_eda['weather_label'] = day_eda['weathersit'].map(weather_labels)

# Menampilkan hasil penambahan kolom
print("Kolom tambahan berhasil dibuat!")

print("\nDataset Per Jam:")
display(hour_eda.head())

print("\nDataset Harian:")
display(day_eda.head())

print("\nKategori waktu:")
print(hour_eda['time_category'].value_counts())
```

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
# ANALISIS RATA-RATA PENYEWAAN BERDASARKAN JAM

# Menghitung rata-rata penyewaan sepeda pada setiap jam
hourly_rentals = (
    hour_eda.groupby('hr')['cnt']
    .mean()
    .reset_index()
)

# Mengurutkan berdasarkan jam
hourly_rentals = hourly_rentals.sort_values('hr')

# Mengidentifikasi jam dengan rata-rata penyewaan tertinggi
peak_hour = hourly_rentals.loc[hourly_rentals['cnt'].idxmax()]

# Mengidentifikasi jam dengan rata-rata penyewaan terendah
lowest_hour = hourly_rentals.loc[hourly_rentals['cnt'].idxmin()]

# Menampilkan tabel rata-rata penyewaan
print("RATA-RATA PENYEWAAN SEPEDA BERDASARKAN JAM")
display(
    hourly_rentals.rename(columns={
        'hr': 'Jam',
        'cnt': 'Rata-rata Penyewaan'
    }).round(2)
)

# Menampilkan jam tertinggi dan terendah
print("\nJAM DENGAN RATA-RATA PENYEWAAN TERTINGGI")
print(f"Jam       : {int(peak_hour['hr']):02d}.00")
print(f"Rata-rata : {peak_hour['cnt']:.2f} sepeda")

print("\nJAM DENGAN RATA-RATA PENYEWAAN TERENDAH")
print(f"Jam       : {int(lowest_hour['hr']):02d}.00")
print(f"Rata-rata : {lowest_hour['cnt']:.2f} sepeda")

# Membuat visualisasi pola penyewaan berdasarkan jam
plt.figure(figsize=(12, 6))

plt.plot(
    hourly_rentals['hr'],
    hourly_rentals['cnt'],
    marker='o',
    linewidth=2
)

plt.title('Rata-rata Jumlah Penyewaan Sepeda Berdasarkan Jam')
plt.xlabel('Jam')
plt.ylabel('Rata-rata Jumlah Penyewaan Sepeda')
plt.xticks(range(24))
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Pertanyaan 2:

```python
# ANALISIS PENYEWAAN BERDASARKAN KONDISI CUACA

# Menghitung rata-rata penyewaan berdasarkan kondisi cuaca
weather_rentals = (
    hour_eda.groupby('weather_label')
    .agg(
        rata_rata_penyewaan=('cnt', 'mean'),
        total_penyewaan=('cnt', 'sum'),
        jumlah_observasi=('cnt', 'count')
    )
    .reset_index()
)

# Mengurutkan berdasarkan rata-rata penyewaan tertinggi
weather_rentals = weather_rentals.sort_values(
    by='rata_rata_penyewaan',
    ascending=False
)

# Mengidentifikasi kondisi cuaca dengan rata-rata tertinggi
best_weather = weather_rentals.iloc[0]

# Mengidentifikasi kondisi cuaca dengan rata-rata terendah
lowest_weather = weather_rentals.iloc[-1]

# Menampilkan hasil analisis
print("RATA-RATA PENYEWAAN SEPEDA BERDASARKAN KONDISI CUACA")
display(weather_rentals.round(2))

print("\nKONDISI CUACA DENGAN RATA-RATA PENYEWAAN TERTINGGI")
print(f"Kondisi cuaca : {best_weather['weather_label']}")
print(f"Rata-rata     : {best_weather['rata_rata_penyewaan']:.2f} sepeda")

print("\nKONDISI CUACA DENGAN RATA-RATA PENYEWAAN TERENDAH")
print(f"Kondisi cuaca : {lowest_weather['weather_label']}")
print(f"Rata-rata     : {lowest_weather['rata_rata_penyewaan']:.2f} sepeda")

# Membuat visualisasi
plt.figure(figsize=(12, 6))

sns.barplot(
    data=weather_rentals,
    x='weather_label',
    y='rata_rata_penyewaan'
)

plt.title('Rata-rata Jumlah Penyewaan Sepeda Berdasarkan Kondisi Cuaca')
plt.xlabel('Kondisi Cuaca')
plt.ylabel('Rata-rata Jumlah Penyewaan Sepeda')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

Pada pagi hari, rata-rata penyewaan mencapai 359,01 sepeda per jam pada pukul 08.00, sedangkan pada sore hari mencapai puncaknya pada pukul 17.00. ## Analisis Lanjutan (Opsional)

```python
# ANALISIS LANJUTAN: MANUAL GROUPING & BINNING

# 1. Menghitung rata-rata penyewaan berdasarkan kategori waktu
time_order = ['Dini Hari', 'Pagi', 'Siang', 'Sore', 'Malam']

time_category_analysis = (
    hour_eda.groupby('time_category', observed=False)['cnt']
    .mean()
    .reindex(time_order)
    .reset_index()
)

time_category_analysis.columns = [
    'Kategori Waktu',
    'Rata-rata Penyewaan'
]

print("RATA-RATA PENYEWAAN BERDASARKAN KATEGORI WAKTU")
display(time_category_analysis.round(2))


# 2. Melakukan binning pada rata-rata penyewaan setiap jam
hourly_demand = hourly_rentals.copy()

hourly_demand['kategori_permintaan'] = pd.qcut(
    hourly_demand['cnt'],
    q=3,
    labels=['Rendah', 'Sedang', 'Tinggi']
)

print("\nPENGELOMPOKAN TINGKAT PERMINTAAN BERDASARKAN JAM")
display(
    hourly_demand.sort_values('hr').round(2)
)


# 3. Merangkum jumlah jam pada setiap tingkat permintaan
demand_summary = (
    hourly_demand.groupby('kategori_permintaan', observed=False)
    .agg(
        jumlah_jam=('hr', 'count'),
        rata_rata_penyewaan=('cnt', 'mean'),
        penyewaan_terendah=('cnt', 'min'),
        penyewaan_tertinggi=('cnt', 'max')
    )
    .reset_index()
)

print("\nRINGKASAN TINGKAT PERMINTAAN")
display(demand_summary.round(2))


# 4. Visualisasi rata-rata penyewaan berdasarkan kategori waktu
plt.figure(figsize=(10, 6))

sns.barplot(
    data=time_category_analysis,
    x='Kategori Waktu',
    y='Rata-rata Penyewaan'
)

plt.title('Rata-rata Penyewaan Sepeda Berdasarkan Kategori Waktu')
plt.xlabel('Kategori Waktu')
plt.ylabel('Rata-rata Jumlah Penyewaan Sepeda')
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Berdasarkan hasil analisis lanjutan menggunakan teknik manual grouping, jumlah penyewaan sepeda berhasil dikelompokkan ke dalam lima kategori waktu, yaitu Dini Hari, Pagi, Siang, Sore, dan Malam. Hasil visualisasi menunjukkan bahwa kategori Sore memiliki rata-rata penyewaan tertinggi, diikuti kategori Siang dan Pagi. Sementara itu, kategori Dini Hari memiliki rata-rata penyewaan paling rendah. Hal ini menunjukkan bahwa aktivitas penyewaan sepeda cenderung lebih tinggi pada periode sore dibandingkan periode waktu lainnya.

- Hasil pengelompokan waktu menunjukkan bahwa rata-rata penyewaan pada kategori Sore mencapai sekitar 375 sepeda, sedangkan kategori Siang berada pada kisaran 240 sepeda. Kategori Pagi memiliki rata-rata sekitar 175 sepeda, diikuti Malam sekitar 150 sepeda. Sementara itu, kategori Dini Hari memiliki rata-rata penyewaan paling rendah, yaitu sekitar 25 sepeda. Kategori Sedang memiliki rentang antara 131,34 hingga 240,95 sepeda per jam, sedangkan kategori Tinggi memiliki rentang antara 251,23 hingga 461,45 sepeda per jam.
